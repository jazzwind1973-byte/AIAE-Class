# Q1. 建立一個包含以下水果庫存的 Series：`Apple: 20`, `Banana: 40`, `Cherry: 15`。
2. 進貨了！將 `Apple` 的數量增加 10。
3. 使用**布林篩選 (Boolean Indexing)** 找出所有大於 20 的數值。
4. 把 `Cherry` 從清單中刪除。
5. 輸出最後的庫存清單。

````python
import pandas as pd

# 1. 建立 Series
inventory = pd.Series({'Apple': 20, 'Banana': 40, 'Cherry': 15})
print(inventory, '\n')

# 2. 更新 Apple 數量
inventory['Apple'] += 10
print('inventory["Apple"] += 10\n', inventory, '\n')


# 3. 使用**布林篩選 (Boolean Indexing)** 找出所有大於 20 的數值。
print('庫存清單 > 20 :\n', inventory[inventory>20], '\n')

# 4. 刪除 Cherry (使用 drop)
inventory = inventory.drop('Cherry')

print("更新後的庫存清單：")
print(inventory)
````

# Q7 我們先建立一個簡單的銷售紀錄 DataFrame：

```python
import pandas as pd

data = {
    'Department': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing', 'Home'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Sales': [50000, 30000, 45000, 20000, 35000, 25000],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South']
}
df = pd.DataFrame(data)

```
1. 根據 `Department`（部門）進行分組。
2. 計算每個部門的**總銷售額 (Sales)**。
3. 根據 `Department` 進行分組。
4. 同時計算每個部門銷售額的**平均值 (mean)**、**最大值 (max)** 以及**員工人數 (count)**。
5. **提示：** 使用 `.agg()` 函數。

````python
import pandas as pd

data = {
    'Department': ['Electronics', 'Clothing', 'Electronics', 'Home', 'Clothing', 'Home'],
    'Employee': ['Alice', 'Bob', 'Charlie', 'David', 'Eve', 'Frank'],
    'Sales': [50000, 30000, 45000, 20000, 35000, 25000],
    'Region': ['North', 'South', 'North', 'South', 'North', 'South']
}
df = pd.DataFrame(data)

# 1. 根據 Department（部門）進行分組。
# 2. 計算每個部門的總銷售額 (Sales)。
dept_sales = df.groupby('Department')['Sales'].sum()

print("各部門總銷售額：")
print(dept_sales)

# 3. 根據 Department 進行分組。
# 4. 同時計算每個部門銷售額的平均值 (mean)、最大值 (max) 以及員工人數 (count)。
dept_stats = df.groupby('Department')['Sales'].agg(['mean', 'max', 'count'])

print("各部門詳細統計：")
print(dept_stats)
````

# Q10  Data Wrangling 中最核心的「資料組合」動作，

你有兩張表，一張是「員工姓名表」，另一張是「員工部門表」。

1. 建立兩個 DataFrame：
* `df_names`: 包含 `ID` (1, 2, 3) 與 `Name` (Alice, Bob, Charlie)。
* `df_depts`: 包含 `ID` (2, 3, 4) 與 `Dept` (HR, IT, Sales)。

```python
import pandas as pd

# 1. 準備資料
df_names = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df_depts = pd.DataFrame({'ID': [2, 3, 4], 'Dept': ['HR', 'IT', 'Sales']})
```
2. 使用 **Inner Join**（內連接）合併這兩張表。
3. 使用 **Left Join**（左連接）合併這兩張表，並觀察結果。
4. 將這兩個 DataFrame 垂直合併成一整年的清單。
5. **要求：** 合併後要重新重置索引（index），避免出現重複的 0, 1。

````python
import pandas as pd

# 1. 準備資料
df_names = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df_depts = pd.DataFrame({'ID': [2, 3, 4], 'Dept': ['HR', 'IT', 'Sales']})

# 2. Inner Join (只保留 ID 相同的交集：2, 3)
inner_result = pd.merge(df_names, df_depts, on='ID', how='inner')

# 3. Left Join (保留左表所有 ID，右表無對應則補 NaN)
left_result = pd.merge(df_names, df_depts, on='ID', how='left')

print("Inner Join 結果：\n", inner_result)
print("\nLeft Join 結果：\n", left_result)

# 4 & 5 垂直合併並重置索引
# ignore_index=True 會捨棄原本的索引並生成 0, 1, 2, 3...
# df_comb = pd.concat([df_names, df_depts], ignore_index=True)
df_comb = pd.merge(df_names, df_depts, on='ID', how='outer')
print(f"\n合併後的年度清單：\n{df_comb}")
````

````python
import pandas as pd

# 1. 準備資料
df_names = pd.DataFrame({'ID': [1, 2, 3], 'Name': ['Alice', 'Bob', 'Charlie']})
df_depts = pd.DataFrame({'ID': [2, 3, 4], 'Dept': ['HR', 'IT', 'Sales']})
print(df_names, '\n\n', df_depts, '\n')
# 2. Inner Join (只保留 ID 相同的交集：2, 3)
inner_result = pd.merge(df_names, df_depts, on='ID', how='inner')
# print('ddd', pd.concat([df_names, df_depts], axis=1, join='inner'))
# 3. Left Join (保留左表所有 ID，右表無對應則補 NaN)
left_result = pd.merge(df_names, df_depts, on='ID', how='left')

print("Inner Join 結果：\n", inner_result)
print("\nLeft Join 結果：\n", left_result)

# 4 & 5 垂直合併並重置索引
# ignore_index=True 會捨棄原本的索引並生成 0, 1, 2, 3...
df_comb = pd.concat([df_names, df_depts], ignore_index=True)

print("\n合併後的年度清單：")
print(df_comb)
````
