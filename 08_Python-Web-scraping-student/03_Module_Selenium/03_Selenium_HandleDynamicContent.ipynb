Interact with elements by Hover 透過滑鼠懸停來互動

```python
import os
import time
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver import ActionChains
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from time import sleep

# 1. 設定 HTML 檔案路徑
file_path = "file://" + os.path.abspath("action_chains_demo.html")

driver = webdriver.Chrome()
try:
    print("開始載入頁面...")
    driver.get(file_path)
    
    # 強制暫停 3 秒，等待頁面渲染 (註：實務上仍建議使用 Explicit Wait 以提高效率)
    time.sleep(3)

    # 2. 定位搜尋輸入框
    # 說明：在進行文字輸入前，必須先精準取得該輸入框的網頁元素 (WebElement)。
    # 主選單
    #<div class="menu" id="menu-hover">
    main_menu = driver.find_element(By.ID, 'menu-hover')

    # 子選單
    # <a href="#" id="link-option-1" onclick="clickSubMenu()">子選單項目 1</a>
    submenu = driver.find_element(By.ID, 'link-option-1')

    action = ActionChains(driver)

    action.move_to_element(main_menu)


    action.click(submenu)

    # 目前都是新增到任務佇列裡

    action.perform()


    # 暫停 5 秒觀察執行結果
    time.sleep(5)

except Exception as e:
    # 捕捉並印出執行過程中發生的任何錯誤
    print(f"執行過程中發生錯誤: {e}")

finally:
    # 5. 釋放資源
    driver.quit()
    print("驅動程式已關閉。")
```
