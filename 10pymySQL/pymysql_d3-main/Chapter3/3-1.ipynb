```python
import pymysql
from configparser import ConfigParser

config = ConfigParser()
config.read('../Chapter1/config.ini')

connection = pymysql.connect(
    host=config.get('DB', 'host'),
    user=config.get('DB', 'user'),
    password=config.get('DB', 'password'),
    port=config.getint('DB', 'port'),
    cursorclass=pymysql.cursors.DictCursor,
)

# 建立資料庫
with connection.cursor() as cursor:
    sql = """
    CREATE DATABASE IF NOT EXISTS chapter3;
    """
    cursor.execute(sql)

    # 建立有時間紀錄的表格
    # 包含 create_at 欄位 預設在寫入時使用現在的時間
    sql = """
        CREATE TABLE IF NOT EXISTS chapter3.users (
            id INT AUTO_INCREMENT PRIMARY KEY,
            name VARCHAR(255) NOT NULL,
            age INT,
            username VARCHAR(255) NOT NULL UNIQUE,
            password VARCHAR(255) NOT NULL,
            create_at TIMESTAMP DEFAULT CURRENT_TIMESTAMP
        );
    """

    cursor.execute(sql)
    cursor.execute("SHOW TABLES FROM chapter3;")
    result = cursor.fetchall()

print(result)
```

```python
from pprint import pprint
with connection.cursor() as cursor:
    sql = """
        INSERT INTO chapter3.users (name, age, username, password)
        VALUES ('Jarvis', 30, 'jarvis_1029', '123456')
    """
    cursor.execute(sql)
    connection.commit()
    cursor.execute("SELECT * FROM chapter3.users;")
    result = cursor.fetchall()
    
pprint(result)
```

```python
from datetime import datetime
# import datetime
# now = datetime.datetime.now()

# 取得當前時間
now = datetime.now()
print(now)

# 取得指定時間的 datetime 物件
data = datetime(2025, 10, 1)
print(data)

# 年/月/日
print(data.year)
print(data.month)
print(data.day)

print(data.date())

# 輸出成指定格式
format_date = data.strftime("%Y/%m/%d")
print(format_date)
format_time = now.strftime("%Y/%m/%d %H時%M分%S秒")
print(format_time)

# 取得民國年
print(f"民國 {datetime.now().year - 1911} 年")
```

```python
# 將指定格式的字串轉換成 datetime 物件
data = "2026/10/07 11點11分"
datetime_obj = datetime.strptime(data, "%Y/%m/%d %H點%M分")
print(datetime_obj)
print(datetime_obj.date())

data = "10/17/2020" # 月/日/年
datetime_obj = datetime.strptime(data, "%m/%d/%Y")
print(datetime_obj)
```

```python
# 取得前一天日期
from datetime import timedelta, datetime
yesterday = datetime.now() - timedelta(days=1)
print(yesterday.date())
```

```python
# ISO format
iso_string = '2025-01-01T12:00:00'

# 轉換成 datetime 物件
result = datetime.fromisoformat(iso_string)
print(result)
```
