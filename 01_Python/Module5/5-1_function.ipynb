```python
# 建立 function 3
def greeting(name):
    greeting_str ="Hello, Good Monimg!"

    return greeting_str + " " + name

result = greeting("Jarvis")
print(result)
result = greeting("Alice")
print(result)
```

```python
# 建立 function 4
def get_total(total, discount):
    if total >= 2000:
        cnt = total // 2000 #計算要折幾次
        total -= cnt*discount

    return total

result = get_total(8000,100)
print(f"結帳金額:{result}元")
result = get_total(6400, 300)
print(f"結帳金額:{result}元")
```

```python
# *args **kwargs

def add_nums(*args, **kwargs):
    print(args)
    print(kwargs)
    total = 0
    for i in args:
        total += i

    return total

total = add_nums(1, 2, 3, 4, name="Jarvis", age=30)
print(total, "Hello", "World")
```
