```python
# 匯入套件
import os

#某個資料夾裡有哪些檔案
files = os.listdir("../Module7")
print(files)

for file in files:
    if file[-3] == ".md" :
        print(file)
        with open(file, "r", encoding="utf-8")as f:
            data = f.read()
        print("="*50)
        print(data)
        print("="*50)

# 檔案路徑
file_path = "./data.txt"

#檢查某個檔案/資料夾是否存在
print(os.path.exists(file_path))
print(os.path.exists("ttt.txt"))

# 如果指定路徑的檔案存在，則執行刪除
filename = "test.txt"
if os.path.exists(filename):
    os.remove(filename)
    print(f"刪除: {filename}")


```

```python
import os
folder = "lesson"

if not os.path.exists(folder):
    os.mkdir(folder) # 建立資料夾

for i in range(1, 11):
    with open(f"{folder}/{i:03}.txt", "w", encoding="utf-8") as f:
        f.write(f"寫入{i:03}檔案")
```
