##### 將 data.txt 的資料分類成另外三個檔案，分別為 cc.txt 、 css.txt 、 c.txt  

```python
with open ("data.txt", "r", encoding="utf-8")as f:
    for line in f:
        if line.strip().endswith(".cc"):
            with open("cc.txt", "a", encoding="utf-8") as out :
                out.write(line)

        elif line.strip().endswith(".css"):
            with open("css.txt", "a", encoding="utf-8") as out :
                out.write(line)

        elif line.strip().endswith(".c"):
            with open("c.txt", "a", encoding="utf-8") as out :
                out.write(line)
    

```
