### `json` 套件

| 用法 | 說明 |
|:--:|:--:|
| |物件轉JSON並寫入文件|
| |物件轉JSON字串|
| |讀取JSON文件|
| |將JSON字串轉成Python物件|

```python
import json

user_data= [
    {
        "name":"Alex",
        "age" : 30,
        "role": "admin"
        "in_school": True,
        "class": None
    },
    {
        "name":"John",
        "age": 30,
        "role":"user"
        "in_school": False,
        "class": "AIPE05"
    }

]
with open("users_data.json", "w", encoding="utf-8")as f:
    f.write(json.dumps(user_data, indent=4))
```

```python
# json.dump(<list, dict>),<file>)
```

```python
with open("users_data.json", "r", encoding="utf-8") as f:
    data = f.read()

print(type(data))
#來源可能不只是檔案!!
data = json.loads(data)
print(type(data))
print(data)
```

```python
web_element = "{ &quot;type&quot; : &quot;static&quot;, &quot;id&quot; : &quot;892415667&quot;, &quot;staticUrl&quot; : &quot;https://stickershop.line-scdn.net/stickershop/v1/sticker/892415667/android/sticker.png?v=1&quot;, &quot;fallbackStaticUrl&quot; : &quot;https://stickershop.line-scdn.net/stickershop/v1/sticker/892415667/android/sticker.png?v=1&quot;, &quot;animationUrl&quot; : &quot;&quot;, &quot;popupUrl&quot; : &quot;&quot;, &quot;soundUrl&quot; : &quot;&quot; }"

web_element = web_element.replace("&quot;", '"')

print(web_element)
print(type(web_element))

web_element = json.loads(web_element)
print(type(web_element))
print(web_element["staticUrl"])
```
