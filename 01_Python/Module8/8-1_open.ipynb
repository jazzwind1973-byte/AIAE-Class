```python
# 一行一行讀取
with open("../Module2/2-1.py", "r", encoding="utf-8") as f:
        content_list = f.readlines()
        #消除/n完成整理
        for line  in content_list :
            line = line.replace("\n", "")
            print(line)
```

```python
# 資料處理
with open("../Module2/2-1.py", "r", encoding="utf-8") as f:
        content_list = f.readlines()

print(content_list)

while "\n" in content_list:
    content_list.remove("\n")
print(content_list)

result = []
for content in content_list:
    result.append(content.strip("\n"))
print(result)
```

##### 寫入檔案 write( )

```python
# 文字內容
text = '''
我這一生如履薄冰，你說我能走到對岸嗎
'''

# 寫入檔案
with open("test.txt", "w", encoding="utf-8")as f:
    f.write(text)


```

```python
# 寫入資料到最後一行 (附加文字在檔案內容最後一行)

text = "喲，這不薄冰哥嗎，今天還沒走到對岸呢"
with open("test.txt", "a", encoding="utf-8")as f:
    f.write(text + "\n")
```

##### 讀取 binary 轉圖片

```python
with open("img.bin", "rb") as f:
    bin_data = f.read()
print(bin_data)

with open("img.jpg", "wb")as f:
    f.write(bin_data)
```
