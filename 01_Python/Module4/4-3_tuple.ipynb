```python
# tuple 初始化: 第一種
nums = (1, 2, 3)
print(nums)
print(type(nums))
```

```python
# tuple 初始化: 第二種
nums = 1, 2, 3
print(nums)
print(type(nums))
```

```python
# 透過指定索引輸出值
nums = (1, 2, 3)
print(nums[1])
```

```python
#多變數取值
#a, b, c = (1, 2, 3)
a, b, c = 100, 20, 35
print(a)
print(b)
print(c)
print(a, b, c)
```

```python
# ( list)可以修改指定索引的值
nums = (1, 2, 3)
nums = list(nums)
print(nums)
```
