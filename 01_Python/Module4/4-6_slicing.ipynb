```python
# 從 6 取到尾
data = "Hello,World!"
print(data[6:])
```

##### 使用負號

```python
# 從倒數第 10 取到倒數第 7（不包含倒數第 7 的元素）
data = "Hello,World!"
print(data[-10:-7])

```

```python
# 從倒數第 6 取到尾
data = "Hello,Word!"
print(data[-6:])
```

```python
# 從頭取到倒數第 7（不包含倒數第 7 的元素）
data = "Hello,World!"
print(data[:-7])
```

```python
# 也可以使用在 list
data = [1, 2, 3, 4, 5]
print(data[2:])
print(data[-1])
```
