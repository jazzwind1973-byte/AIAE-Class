```python
# 初始化一個 list
nums = [1, 2, 3, 4]
print(nums)
print(type(nums))
```

```python
# 新增元素在 list 尾端
nums = [1, 2, 3, 4]
nums.append(5)
print(nums)
```

```python
# 修改索引位置的元素
nums = [1, 2, 3, 4]
print(nums[2])

nums[2] = 100
print(nums)
```

```python
# 刪除索引位置的元素
nums = [1, 2, 3, 4]
nums.pop(2)
print(nums)
```

```python
# 新增元素 9 在指定索引 1，其餘元素往後移
nums = [1, 2, 3, 4]
nums.insert(1, 9)
print(nums)

```

```python
# 移除指定的值
nums = [1, 2, 3, 4]
nums.remove(3)
print(nums)
```

```python
# 計算 list 內的某個值 共有幾個
nums = [1, 2, 3, 4]
for _ in range(5):
    nums.append(9)

print(nums)
print(nums.count(9))
```

```python
data = [1, 2, 3, 4, 9, 9, 9, 9, 9,]

while 9 in data :
    data.remove(9)
print(data)
```

```python
#移除9直到剩一個
while data.count(9) != 1:
    data.remove(9)

print(data)
```

```python
data = [1, 2, 3, 4, 9, 9, 9, 9, 9,]
# 移除 9 直到剩一個
for _ in range(data,count(9)):
    data.remove(9)

print(data)
```

### list 搭配 for 迴圈

```python
# 建立一個 1~10 的 list
nums = []
for i in range(1,11):
    nums.append(i)

print(nums)

```

```python
cnt = 0
for num in data:
    if num == 8:
        cnt += 1
        print(f"8 在 index: {cnt}")
    # print(num, end=", ")
```

```python
# enumerate() 同時取得 index 與 value
for index, value in enumerate(data):
    if value == 8:
        print(f"{value} 在 index: {index}")
        
```

```python
#查看 list 的長度
nums = [1, 2, 3, 4]
print(len(nums))
```

```python
#list的合併
nums_1 = [1, 2, 3]
nums_2 = [4, 5, 6]

result = nums_1 + nums_2
print(result)

nums_1.extend(nums_2)
print(nums_1)
```

```python
# 錯誤示範
nums_1 = [1, 2, 3]
nums_2 = [4, 5, 6]

result = nums_1 + nums_2
print(result)

nums_1.append(nums_2)
print(nums_1)
```

### List comprehension 列表推導式
`new_list = [expression for item in iterable [if condition]]`  
  
`list的名稱 = [表達式 for 元素 in 迭代對象 [if 判斷式(...的條件)]]`

```python
nums = []
for i in range(1,11):
    nums.append(i)
print(nums)
```

```python
nums = [i for i in range(1, 11)]
print(nums)
```

```python
result = []
for num in nums:
    if num % 2 == 0:
        result.append(num)

print(result)
```

```python
result = [num for num in nums if num % 2 == 0]
print(result)
```
