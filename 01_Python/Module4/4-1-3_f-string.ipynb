```python
# f-string (又稱作 formatted string literals, version >= 3.6)
name = "Jarvis"
age = 30

msg = f"{name} is {age} years old."
print(msg)
```

```python
# 靠左對齊、靠右對齊
name = "Jarvis"
age =30
msg =f"Name [{name:10}] Age:[{age:10}]"
print(msg)

msg = f"Name:"
```

```python
# 靠左對齊、靠右對齊，字數不足，補「ㄏ」
# (沒寫補什麼，預設空格)
name = "Jarvis"
age = 30
msg = f"Name: [{name:*<10}] age:[{age:*>10}]"
print(msg)
```

```python
# 指定文字長度上限 (只有文字，才在格式化字串中加入「.」來限定字串長度)
name = "Jarvis"
age = 30
msg = f"Name:[{name:.2}]Age:[{age:^10}]"
print(msg)
```

```python
#指定小數點位數
result = 3.15487548
print(f"{result:.2f}")
```

```python
#看金額
total = 100000000
print(f"{total:,}")
```
