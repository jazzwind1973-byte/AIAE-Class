|string方法| 功能 |
|:---:|:---:|
| string.replace(<old>,<new>)| 替換字串 |
| | 去除兩側空格 |
| | 字串小寫變大寫 |
| | 字串大寫變小寫 |

```python
# 字串變數初始化
name = "Alex"
print(name)
```

```python
# 替換字串
'''
str.replace("old", "new")
將 string 中的 str1 替換成 str2
'''
name = "Alex"
new_name = name.replace("ex", "len")
print(name)
print(new_name)
```

```python
# 去除兩側空格
'''

去除字串 string 左、右兩邊的空格
'''
data = " _Alice_"
print(data)

result = data.strip().strip
```

```python
# 字串小寫變大寫
'''

將字串 string 裡的字母全部改成小寫
'''

name = "JAVIS"
result = name.lower()
print(result)


'''

將字串 string 裡的字母全部改成大寫
'''
ispan_class= "aipe05"
result = ispan_class.upper()
print(result)

```

```python
# 字串分割成 list
greeting = "hello world! hello John"

result = greeting.split("!")
print(result)
```

```python
# 字串合併
greet_str = "Good Moning"
user_name = "Jarvis"
result = greet_str + user_name
print(result)
```

```python
# 索引
data = "Hello, World"
print(data[0]) #用中括號取得index
print(data[4])
```
