```python
'''
改寫 fruit_dict 每個水果需要 price, quantity/amount, color
apple 有 綠色 和 紅色
orange 有 橘色
tomato 有 紅色
'''
fruit_dict = {

    "apple": {
        "price":30,
        "quantity": 50,
        "color":["green", "red"]        
    },

    "orange": {
        "price": 50,
        "quantity": 100,
        "color":["orange"]

    },

    "tomaro": {
        "price": 20,
        "quantity": 200,
        "color":["red"]
    
    },

}

print(fruit_dict)

fruit_dict2 = {
    "kiwi": {
        "price": 100,
        "quantity": 20,
        "color": ["green"]
    }
}
```

##### 另外的 print 方式 `pprint` (pretty-print) 

```python
# 從 Module 引入 function
from pprint import pprint
pprint(fruit_dict, sort_dicts=False)
```

```python
'''
新增顏色
apple 有新的品種 顏色要有黃色
tomato 有綠色
'''
if "yellow" not in fruit_dict["apple"]["color"]:
    fruit_dict["apple"]["color"].append("yellow")

if "green" not in fruit_dict["tomato"]["color"]:
    fruit_dict["tomato"]["color"].append("green")

pprint(fruit_dict)
```
