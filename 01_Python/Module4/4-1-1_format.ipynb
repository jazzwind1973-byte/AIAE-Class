```python
# 靠左對齊 (維持 10 個字元長度，向左對齊)
msg = "[%-10s]" % ("Hello")
print(msg)
```

```python
# 指定浮點數位數
msg = "%0.3f" % (3.14569385)
print(msg)
```

```python
# 指定文字長度上限 (只有文字，才在格式化字串中加入「.」來限定字串長度)
msg = "[%.3s]" % ("hello")
print(msg)
```
