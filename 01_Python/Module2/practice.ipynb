百貨公司周年慶 消費金額超過 2000 要折 200  

請撰寫一支程式  
計算客戶消費的結帳金額

```python
x = int(input("請輸入消費金額"))
total = x
if x >=2000 :
    total-= 200
    print(total)
```

滿2000折200 滿4000 折400 滿6000折600...以此類推

```python
x = int(input("請輸入消費金額"))
total = x
if x >=2000 :
    total-= 200*(x // 2000)
    print(total)
```
