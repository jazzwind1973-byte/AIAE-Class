算術運算子

```python
num = 20

print(num * 5)
print(2 ** 5)
print(num & 3) # 6餘2 ->相除取餘數
print(num // 6) # 3除2 -> 相除取整數
```

```python
score = 50
passing_score = 60 
print(score < passing_score)
if score < passing_score: #如果為 True 就執行底下區域的程式碼
    score = score + 10 # score +=10
    print("加分")
#反縮排 shift + Tab
print(score)
```

邏輯運算子 and / or / not

```python
name = "Jarvis"
age = 30

if name == "Jarvis" and age == 30 and age >18:
    print("and 條件全部都為 True 才為 True")
if name == "Jarvis" or age == 50:
    print("or 其中一個為 True 就是 True")
if not name == "Alex":
    print("not Fasle才執行")
```
