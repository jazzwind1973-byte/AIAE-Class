##### try...except...

```python
# 可以連續定義數個例外處理機制
# 最符合的例外事件類別會先拋出
try:

    num = [1, 2, 3]
    print(num[3])
    print(len(nums))
except:
    print("發生錯誤")
```

```python
try:
    print(x)
    nums = [1, 2, 3]
    print(num[3])
    print(len(nums))
except:
    print("發生錯誤")
```

```python
try:
    print("1"+ 1)# TyperError
    print(x)# NameError
    nums = [1, 2, 3]
    print(nums[3])
    print(len(nums))
except IndexError: #捕捉特定的Error
    print("Index 發生錯誤")
    print("發生錯誤")
except NameError:
    print("發生 NameError")
except Exception:
    print("發生其他錯誤")
```

```python
try:
    num = [1, 2, 3]
    print(num[3])
    print(len(nums))
except Exception as err:
    print(f"發生錯誤:(err)")
```
