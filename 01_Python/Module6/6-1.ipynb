```python
# 類別名稱
# 類別屬性 (Class Attribute)
class Car:
    wheel = 4

    def __init__(self, brand, model):
        self.brand = brand #實例屬性AAA
        self.model = model
    def info(self):
        return f"{self.brand} has Model {self.model} has {self.wheel}wheels."

print(Car.wheel)# 取得類別屬性的值
#建立實例
car1 = Car("Benz", "C300")
print(f"car1:{car1.brand}")
print(f"car1:{car1.model}")
print(f"car1:{car1.wheel}")
print(car1.info())


car2= Car("Toyota", "Altis")
print(f"car2:{car2.brand}")
print(f"car2:{car2.model}")
print(f"car2:{car2.wheel}")
```
