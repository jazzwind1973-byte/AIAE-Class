##### break

```python
'''
當偵測到字母 t 時，就會強制結束迴圈
'''
for i in "contents":
        if i == "t":
            break
        print(i)
```

##### continue

```python
'''
當偵測到字母 t 時，
會跳過本次迴圈剩下的程式碼 print(char)，
但不會結束迴圈，仍然會進入下一圈繼續執行
'''
for i in "contents":
        if i == "t":
            continue
        print(i)

```

### 小試身手
計算 1 到 100 之間的和，如果值超過500就停止執行

```python
cnt = 0
total =0

while True :
    cnt += 1
    total += cnt
    print(total,end="," )
    if (total)>500 :
        break
```

```python
cnt = 0
for i in range(1, 101):
    cnt+= 1
    print(cnt,end=", ")
    if cnt >500 :
        break
```

計算 1 到 100 之間的奇數之和，但要跳過所有被 5 整除的數

```python
total = 0
for i in range(1, 101):
    if i % 2 == 0 or i % 5 == 0:
        continue
    total += i
    #print(1, end=", ")
print(total)
```
