條件語句 if-elif-else

```python
'''
條件判斷
if ...
if ... else ...
if ... elif ... else ...

'''
```

##### if

```python
age = 30
if age >= 18
    print("已滿18歲")
```

##### if...else...

```python
age = 12
if age >= 18:
    print("已滿18歲")
else:
    print("未滿18歲")
```

##### if...elif... (else)

```python
score = int(input("請輸入分數"))
if score == 100:
        print("恭喜你滿分")
elif 60 <score <= 100:
        print("差一點")
else:
        print("再接再厲")
```
