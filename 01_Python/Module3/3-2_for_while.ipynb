| 用法 | 說明 |
|:---:|:---:|
| `range(m)` | 走訪 0 到 m-1 的數字 |
| `range(n, m)` | 走訪 n 到 m-1 的數字  |
| `range(n, m, step)` | 以step為間距，走訪 n 到 m-1 的數字 |

##### for

```python
for i in "content":
    print(i, end=", ")
```

```python
# range(m-1)
for i in range(5):
    print("Hello world!s")
```

```python
# range(n, m-1, step)
# range(n, m-1)
for i in range(1, 101, 2):
    print(i, end=", ")
```

### 小試身手 for  
請產出 a 開頭的 n 個偶數(含 a 本身)  
Ex:  
a = 5, n = 3  
Output : 6 8 10

a = 4, n = 5  
Output : 4 6 8 10 12

```python

#除2餘0->偶數
#除2餘1->奇數

a=5
n=3

if a % 2 ==0:
    for i in range(n):
        print(a + i*2, end=" ")

else :
    a +=1
    for i in range(n):
        print(a + i*2, end=" ")


    
```
